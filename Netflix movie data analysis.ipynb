# Cateforizing Vote_Average Column

We would cut the Vote_Average values and make 4 categories : Popular Average Below_Aug
Not_Popular to describe it more using categorize_col() function provided above

```python
def categorize_col(df, col, labels):

    edges = [df[col].describe()['min'],
             df[col].describe()['25%'],
             df[col].describe()['50%'],
             df[col].describe()['75%'],
             df[col].describe()['max']]
    df[col] = pd.cut(df[col], edges, labels = labels, duplicates = "drop")    # cut - Helps in categorizing the function
    return df
```

```python
labels = ['not_popular', 'below_avg', 'average', 'popular']

categorize_col(df, 'Vote_Average', labels)

df['Vote_Average'].unique()
```
